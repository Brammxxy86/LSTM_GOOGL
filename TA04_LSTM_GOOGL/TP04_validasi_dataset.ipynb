```
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```
df = yf.download('GOOGL', period='5y', interval='1d', auto_adjust=True, progress=False)
if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)
df = df.reset_index()
df['Date'] = pd.to_datetime(df['Date'])
df = df[['Date', 'Close']].dropna().sort_values('Date')
df.set_index('Date', inplace=True)
df.head()
```

```
print('Jumlah observasi:', len(df))
print('Periode:', df.index.min(), 's.d.', df.index.max())
print('Missing value:', df['Close'].isna().sum())
```

```
plt.figure(figsize=(12, 5))
df['Close'].plot()
plt.title('Harga Penutupan Saham Alphabet (GOOGL)')
plt.xlabel('Tanggal')
plt.ylabel('Harga Penutupan (USD)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
